```
import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit

# 1. Separate the known answers from the measurements.
labels = df["Label"].astype("string").str.strip()

if not labels.isin(["BENIGN", "PortScan"]).all():
    raise ValueError("Missing or unexpected labels need investigation.")

features = df.drop(columns="Label").apply(
    pd.to_numeric, errors="coerce"
)

# 2. Treat infinite values as missing.
features = features.replace([np.inf, -np.inf], np.nan)

# 3. Remove the extra header column only if it is identical.
if (
    "Fwd Header Length.1" in features.columns
    and "Fwd Header Length" in features.columns
    and features["Fwd Header Length.1"].equals(
        features["Fwd Header Length"]
    )
):
    features = features.drop(columns="Fwd Header Length.1")
    print("Removed an identical extra header column.")

# 4. Keep one copy of each identical measurement-and-label row.
prepared = pd.concat(
    [features, labels.rename("Label")], axis=1
)
before = len(prepared)
prepared = prepared.drop_duplicates().reset_index(drop=True)

print("Repeated rows removed:", before - len(prepared))
print("Records remaining:", len(prepared))

# X = measurements; y = answer: 0 normal, 1 PortScan.
X = prepared.drop(columns="Label")
y = prepared["Label"].map({"BENIGN": 0, "PortScan": 1}).astype(int)

# 5. Give identical measurement rows the same group ID.
# This also keeps identical inputs with different labels together.
groups = pd.util.hash_pandas_object(X, index=False).to_numpy()

first_split = GroupShuffleSplit(
    n_splits=1, test_size=0.30, random_state=42
)
train_idx, remaining_idx = next(
    first_split.split(X, y, groups=groups)
)

second_split = GroupShuffleSplit(
    n_splits=1, test_size=0.50, random_state=42
)
val_relative, test_relative = next(
    second_split.split(
        X.iloc[remaining_idx],
        groups=groups[remaining_idx]
    )
)

val_idx = remaining_idx[val_relative]
test_idx = remaining_idx[test_relative]

X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
X_test, y_test = X.iloc[test_idx], y.iloc[test_idx]

# 6. Inspect the actual sizes and class counts.
for name, measurements, answers in [
    ("Training", X_train, y_train),
    ("Validation", X_val, y_val),
    ("Testing", X_test, y_test),
]:
    print(f"\n{name}: {measurements.shape}")
    print(
        answers.value_counts().rename(
            index={0: "BENIGN", 1: "PortScan"}
        )
    )
```

```
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# 1. Start from the original data without changing df
data = df.copy()
data.columns = data.columns.str.strip()

# Clean labels; never guess a missing answer
data["Label"] = data["Label"].astype("string").str.strip()
data = data[
    data["Label"].isin(["BENIGN", "PortScan"])
].copy()

# 2. Convert feature columns to numbers
feature_names = data.columns.drop("Label")

data[feature_names] = data[feature_names].apply(
    pd.to_numeric, errors="coerce"
)

# Infinity cannot be used by the model: treat it as missing
data[feature_names] = data[feature_names].replace(
    [np.inf, -np.inf], np.nan
)

print("Missing numeric cells before filling:",
      int(data[feature_names].isna().sum().sum()))

# 3. Remove identical labelled rows for this benchmark
before = len(data)
data = data.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", before - len(data))
print("Remaining records:", len(data))

# 4. Separate measurements from answers
X = data.drop(columns="Label")
y = data["Label"].map({"BENIGN": 0, "PortScan": 1}).astype(int)

# Keep identical feature rows in the same partition,
# including any rows that have conflicting labels
groups = pd.util.hash_pandas_object(X, index=False).to_numpy()

splitter = GroupShuffleSplit(
    n_splits=1, test_size=0.30, random_state=42
)

train_idx, remaining_idx = next(
    splitter.split(X, y, groups=groups)
)

remaining_splitter = GroupShuffleSplit(
    n_splits=1, test_size=0.50, random_state=42
)

val_relative_idx, test_relative_idx = next(
    remaining_splitter.split(
        X.iloc[remaining_idx],
        y.iloc[remaining_idx],
        groups=groups[remaining_idx]
    )
)

val_idx = remaining_idx[val_relative_idx]
test_idx = remaining_idx[test_relative_idx]

X_train = X.iloc[train_idx].copy()
X_val = X.iloc[val_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_val = y.iloc[val_idx].copy()
y_test = y.iloc[test_idx].copy()

for name, target in [
    ("Training", y_train),
    ("Validation", y_val),
    ("Testing", y_test)
]:
    assert target.nunique() == 2, f"{name} needs both classes"

# 5. Identify uninformative columns using TRAINING data only
# This also removes columns that are completely missing
drop_columns = X_train.columns[
    X_train.nunique(dropna=True) <= 1
].tolist()

# Remove the repeated header feature if it matches in training
duplicate_header = "Fwd Header Length.1"

if (
    duplicate_header in X_train.columns
    and "Fwd Header Length" in X_train.columns
    and X_train[duplicate_header].equals(
        X_train["Fwd Header Length"]
    )
):
    drop_columns.append(duplicate_header)

drop_columns = list(dict.fromkeys(drop_columns))

X_train = X_train.drop(columns=drop_columns)
X_val = X_val.drop(columns=drop_columns)
X_test = X_test.drop(columns=drop_columns)

feature_columns = X_train.columns.tolist()

# 6. Learn medians from training data, then fill all three sets
imputer = SimpleImputer(strategy="median")

X_train_filled = imputer.fit_transform(X_train)
X_val_filled = imputer.transform(X_val)
X_test_filled = imputer.transform(X_test)

# 7. Learn scaling from training data, then apply to other sets
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_filled)
X_val_scaled = scaler.transform(X_val_filled)
X_test_scaled = scaler.transform(X_test_filled)

# 8. Verify the processed measurements
for values in [
    X_train_scaled, X_val_scaled, X_test_scaled
]:
    assert np.isfinite(values).all()

print("\nRemoved columns:", drop_columns)
print("Features retained:", len(feature_columns))

for name, target in [
    ("Training", y_train),
    ("Validation", y_val),
    ("Testing", y_test)
]:
    print(f"\n{name}: {len(target)} records")
    print(
        target.map({0: "BENIGN", 1: "PortScan"})
        .value_counts()
    )

print("\nPreprocessing complete: no missing or infinite values.")
```

```
import matplotlib.pyplot as plt
import pandas as pd

# 1. Class distribution: training data only
class_counts = (
    y_train.map({0: "BENIGN", 1: "PortScan"})
    .value_counts()
    .reindex(["BENIGN", "PortScan"])
)

ax = class_counts.plot(
    kind="bar",
    color=["steelblue", "tomato"],
    figsize=(6, 4)
)

ax.set_title("Training Data: Normal vs PortScan")
ax.set_xlabel("Traffic label")
ax.set_ylabel("Number of records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("Training class percentages:")
print((class_counts / class_counts.sum() * 100).round(2))

# 2. Summarize a few useful measurements before scaling
selected_features = [
    "Flow Duration",
    "Total Fwd Packets",
    "Total Backward Packets",
    "Flow Bytes/s",
    "SYN Flag Count"
]

display(X_train[selected_features].describe().round(2))

# 3. Examine feature distributions
# Clip only the plotted values so extreme values do not hide
# the main distribution. The actual training data is unchanged.
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, feature in zip(
    axes,
    ["Flow Duration", "Total Fwd Packets", "Flow Bytes/s"]
):
    values = X_train[feature].dropna()
    lower = values.quantile(0.01)
    upper = values.quantile(0.99)

    values.clip(lower=lower, upper=upper).hist(
        bins=40, ax=ax, color="steelblue"
    )

    ax.set_title(feature)
    ax.set_xlabel("Value (plot clipped at 1st/99th percentiles)")
    ax.set_ylabel("Number of records")

plt.tight_layout()
plt.show()
```

```
# These measurements should not be negative
nonnegative_features = [
    "Flow Duration",
    "Flow Bytes/s",
    "Flow Packets/s",
    "Fwd Packets/s",
    "Bwd Packets/s"
]

# Apply the same cleaning rule to all three sets
for name, frame in [
    ("Training", X_train),
    ("Validation", X_val),
    ("Testing", X_test)
]:
    columns = [
        column for column in nonnegative_features
        if column in frame.columns
    ]

    invalid = frame[columns] < 0

    print(
        f"{name}: {int(invalid.sum().sum())} "
        "negative measurements converted to missing"
    )

    frame[columns] = frame[columns].mask(invalid)

# Check whether cleaning left any training column entirely missing
assert not X_train.isna().all().any(), (
    "A training column is entirely missing; review it before continuing."
)

# Recalculate medians using training data ONLY
imputer = SimpleImputer(strategy="median")

X_train_filled = imputer.fit_transform(X_train)
X_val_filled = imputer.transform(X_val)
X_test_filled = imputer.transform(X_test)

# Recalculate scaling using training data ONLY
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_filled)
X_val_scaled = scaler.transform(X_val_filled)
X_test_scaled = scaler.transform(X_test_filled)

# Verify the data that the model will actually receive
for name, values in [
    ("Training", X_train_scaled),
    ("Validation", X_val_scaled),
    ("Testing", X_test_scaled)
]:
    print(
        f"{name}: missing = {np.isnan(values).sum()}, "
        f"infinite = {np.isinf(values).sum()}"
    )
    assert np.isfinite(values).all()
```
